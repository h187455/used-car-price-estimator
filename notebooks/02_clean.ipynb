```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/Used_Cars_Cleaned.csv")

n0 = len(df)

# Drop useless identifiers
df = df.drop(columns=["id", "posting_date"])

# Filter absurd values (justify each threshold in the report)
df = df[df["price"] >= 500]
df = df[df["price"] <= 150000]
df = df[df["year"] >= 1990]
df = df[df["odometer"] <= 400000]

# Parse "8 cylinders" -> 8.0
df["cylinders"] = df["cylinders"].str.extract(r"(\d+)").astype(float)

# Collapse high-cardinality categoricals
top_models = df["model"].value_counts().head(30).index
df["model"] = df["model"].where(df["model"].isin(top_models), "other")

# Drop columns we won't use
df = df.drop(columns=["region"])  # keep 'state' instead

df.to_csv("../data/cleaned.csv", index=False)
print(f"{n0} -> {len(df)} rows (dropped {n0 - len(df)})")
```
